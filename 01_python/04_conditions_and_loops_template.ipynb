## 1. The `if` statement and indentation
An `if` statement executes its block when its condition is truthy.
In the examples below, comparisons produce Boolean values: `True` or `False`.

- End the condition line with a colon (`:`).
- Indent the block consistently; use four spaces per indentation level.
- Statements at the same indentation level belong to the same block.
- An unindented statement after the block is not controlled by that `if`.
- Use `==` to compare values, not the assignment operator `=`.

### Two alternatives: `if` and `else`
`else` has no condition. Its block runs when the `if` condition is false.
Exactly one of the two blocks runs.

## 2. Multiple branches with `elif` and `else`
`elif` introduces another condition. Python checks an `if`/`elif` chain from top
to bottom and executes only the first matching block. An optional `else` handles
all remaining cases.

Separate `if` statements are independent: more than one of their blocks can run.
Use an `if`/`elif` chain when the choices should be mutually exclusive.

### An `if`/`elif` chain without `else`

## 3. Conditional expressions
A conditional expression selects one of two values:

```python
value_when_true if condition else value_when_false
```

Use it for a short, readable choice of value. Prefer a regular `if` statement
when a branch needs several actions. Only the selected value expression is evaluated.

## 4. `match-case` statements
`match-case` selects a block by matching a value against patterns. It requires
**Python 3.10 or newer**. This lesson uses literal string patterns.

- Put the value to inspect after `match`.
- Indent each `case` under `match`, and its body under `case`.
- Quote string literals, such as `case "standby":`.
- Only the first matching case runs; no `break` is needed.
- Put `case _:` last as the default for any unmatched value.

An unquoted name in a pattern generally captures a value; it is not a comparison
with a string of that name.

## 5. `while` loops
A loop repeats a block of code. Each repetition is called an **iteration**.

A `while` loop checks its condition **before** each iteration and repeats while
that condition is truthy. If the condition is false at the beginning, the body
runs zero times.

Make sure the condition can eventually become false, or provide a deliberate
exit with `break`. Otherwise, the loop may continue indefinitely.

## 6. Loop control: `break` and `continue`
| Statement | Effect |
| --- | --- |
| `break` | Exit the innermost enclosing loop immediately |
| `continue` | Skip the remaining body of the current iteration and proceed to the next one |

Both statements can be used in `while` and `for` loops.
In a `while` loop, `continue` returns execution to the condition check.

## 7. `for` loops and `range()`
A `for` loop processes items from an **iterable**, one item at a time.
Lists, strings, and ranges are examples of iterables.

Unlike a counter-based `while` loop, a `for` loop advances to the next item automatically.
An empty iterable produces no iterations.

### Generating indexes with `range(stop)`
`range(stop)` starts at `0`, advances by `1`, and excludes `stop`.
`range(5)` therefore supplies `0`, `1`, `2`, `3`, and `4`.
A range represents an integer sequence without creating a list of all its values.

### Selecting positions with `range(start, stop, step)`
| Form | Meaning |
| --- | --- |
| `range(stop)` | Start at 0, use step 1, exclude stop |
| `range(start, stop)` | Use the given start, step 1, exclude stop |
| `range(start, stop, step)` | Use the given start and step, exclude stop |

Arguments must be integers, and `step` must not be zero. A negative step counts
down; the start and stop must allow movement in that direction for a nonempty range.

## Summary
| Construct | Use |
| --- | --- |
| `if` | Run a block when a condition is truthy |
| `elif` / `else` | Select among alternative branches |
| Conditional expression | Choose between two values |
| `match-case` | Select a block using patterns; use `_` as the default |
| `while` | Repeat while a condition remains truthy |
| `break` | Exit the innermost enclosing loop |
| `continue` | Skip the rest of the current iteration |
| `for` | Process items from an iterable |
| `range()` | Supply an integer sequence with an excluded stop |
| `enumerate()` | Supply a counter together with each item |

### Common pitfalls
- Missing a colon or using inconsistent indentation.
- Using several independent `if` statements when only one branch should run.
- Placing a broad condition before a more specific one in an `elif` chain.
- Forgetting to update a `while` counter, especially before `continue`.
- Converting the `stop` command to a number before checking it.
- Expecting `range()` to include its stop value.
- Running `match-case` on Python older than 3.10.
