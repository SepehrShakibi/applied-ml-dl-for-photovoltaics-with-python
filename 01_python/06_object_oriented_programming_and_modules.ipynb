## 1. The idea of object-oriented programming
Object-oriented programming (OOP) groups related **data** and **behavior**.
A **class** describes a type of object; an **object**, also called an **instance**,
is a particular example of that class.

Think of a mobile phone:

| Concept | Meaning | Phone example |
| --- | --- | --- |
| Class | A blueprint for related objects | A mobile phone design |
| Object | A particular instance | One person's phone |
| Attribute | Data associated with an object | Model, color, storage capacity, battery status |
| Method | A function defined in a class | Making a call, taking a photo, playing music |

Objects created from the same class share its defined behavior but can store
different attribute values. For example, two phones can have different colors
while supporting the same calling method.

The following examples apply this idea to photovoltaic components.

## 2. Defining a class and creating an object
Use `class`, a class name, a colon, and an indented body. Class names conventionally
use `PascalCase`, such as `SolarPanel`; variables, attributes, and methods use
`snake_case`.

`pass` is a placeholder statement that does nothing. It allows an otherwise empty
class body while we introduce the structure.

Calling `SolarPanel()` creates an instance. The variable `basic_panel` refers to it.

```python
class SolarPanel:
    pass


basic_panel = SolarPanel()
print("Object type:", type(basic_panel).__name__)
print("Is a SolarPanel:", isinstance(basic_panel, SolarPanel))
```

## 3. Initializing objects with external data
`__init__` initializes an instance when it is created. It is often called the
constructor in introductory material; technically, it initializes the object
rather than creating it.

- `self` refers to the current instance. It is the conventional first parameter
  of an instance method.
- Python supplies `self` automatically when calling an instance method through an object.
- Other parameters receive values supplied by the caller.
- `self.model = model` stores a parameter's value as an instance attribute.
- Access an attribute with dot notation, such as `inverter.model`.

### Create a photovoltaic inverter
The caller supplies the model and system status. External data here means data
passed from outside the class; it does not have to come from `input()` or a file.

```python
class Inverter:
    """Store identifying information and status for an example inverter."""

    def __init__(self, model: str, status: str) -> None:
        self.model = model
        self.status = status


inverter_model = "Example Inverter A"
system_status = "Online"

inverter = Inverter(model=inverter_model, status=system_status)
print("Model:", inverter.model)
print("Status:", inverter.status)
```

### Each instance stores its own data
Creating a second inverter does not overwrite the first inverter's instance
attributes. Both objects belong to the same class but have separate state.

The type hints document expected types; they do not validate the supplied data.

```python
backup_inverter = Inverter(model="Example Inverter B", status="Standby")

print("First inverter:", inverter.model, "—", inverter.status)
print("Second inverter:", backup_inverter.model, "—", backup_inverter.status)
print("Same object:", inverter is backup_inverter)
```

## 4. Attributes and methods
A **method** is a function defined inside a class. An instance method can use
`self` to read or modify that object's attributes.

We now replace the introductory `SolarPanel` definition with a version that stores
a model and status and provides a method to display them.

`display_info()` prints a report and implicitly returns `None`; it does not return
the report as a string.

```python
class SolarPanel:
    """Represent a panel with a model and an illustrative operating status."""

    def __init__(self, model: str, status: str) -> None:
        self.model = model
        self.status = status

    def display_info(self) -> None:
        """Display the current model and status."""
        print(f"Panel model: {self.model}")
        print(f"Panel status: {self.status}")
```

### Inspect attributes and call a method
Read an attribute without parentheses. Call a method with parentheses.
In `monitored_panel.display_info()`, Python passes `monitored_panel` as `self`.

```python
monitored_panel = SolarPanel(model="Example Panel A", status="Active")

print("Model attribute:", monitored_panel.model)
print("Status attribute:", monitored_panel.status)
print("\nMethod output:")
monitored_panel.display_info()
```

### Change an attribute from outside the class
These attributes are public and can be reassigned through the object.
The next method call reads the updated value, not a copy of the original status.
This simple class does not restrict which status strings are allowed.

```python
monitored_panel.status = "Under inspection"
monitored_panel.display_info()
```

## 5. Inheritance
Inheritance lets a **child class** reuse functionality from a **parent class**.
Use it when the relationship makes sense: a solar panel **is an** electrical component.

Syntax: `class ChildClass(ParentClass):`.

A child can reuse methods and add its own attributes and methods.
Instance attributes are created when initialization code runs; they are not
copied from a separate parent object.

### Define a general electrical component
The parent stores a model and provides a method that displays it.

```python
class ElectricalComponent:
    """Represent a general electrical component."""

    def __init__(self, model: str) -> None:
        self.model = model

    def display_info(self) -> None:
        """Display the component model."""
        print(f"Component model: {self.model}")
```

### Extend the parent with a solar panel
This section redefines `SolarPanel` as a child of `ElectricalComponent`.

`super().__init__(model)` runs the parent initialization logic for this instance,
creating `self.model`. The child then adds `rated_power_w` and a new method.
Without that call, the child's own `__init__` would not automatically run the
parent's initializer.

Rated power is an illustrative specification, not a measurement of current output.

```python
class SolarPanel(ElectricalComponent):
    """Extend an electrical component with a panel power rating."""

    def __init__(self, model: str, rated_power_w: float) -> None:
        super().__init__(model)
        self.rated_power_w = rated_power_w

    def display_rating(self) -> None:
        """Display the illustrative rated power in watts."""
        print(f"Rated power: {self.rated_power_w:.1f} W")
```

```python
rated_panel = SolarPanel(model="Example Panel B", rated_power_w=400.0)

# Reuse a method defined in the parent class.
rated_panel.display_info()

# Use the additional attribute and method from the child class.
print("Rated power attribute:", rated_panel.rated_power_w)
rated_panel.display_rating()

print("Is a SolarPanel:", isinstance(rated_panel, SolarPanel))
print("Is an ElectricalComponent:", isinstance(rated_panel, ElectricalComponent))
```

## 6. Abstraction and importing from a module
Abstraction describes **what an object must provide** while allowing individual
classes to decide **how to provide it**.

An abstract base class defines a common structure for related classes:
- An **abstract method** declares an operation that a concrete child must implement.
- An **abstract property** declares attribute-style access that a concrete child must provide.
- A class with unimplemented abstract members cannot be instantiated.

### Import the tools from `abc`
A module organizes reusable Python definitions. Python's standard-library `abc`
module provides tools for abstract base classes; no package installation is needed.

`from abc import ABC, abstractmethod` makes these two names available directly
in the notebook. `ABC` is a base class; `abstractmethod` marks required operations.

```python
from abc import ABC, abstractmethod
```

### Define the abstract interface
`EnergyComponent` requires a method named `calculate_output` and a property named
`component_name`.

A decorator is written with `@` above a definition to modify its behavior.
Here, `@property` provides attribute-style access to a getter method, and
`@abstractmethod` marks a member as abstract.

For the abstract property, put **`@property` above `@abstractmethod`**.
The `pass` statements are placeholders; concrete behavior belongs in the child class.

```python
class EnergyComponent(ABC):
    """Define the interface for components that calculate output power."""

    @abstractmethod
    def calculate_output(self) -> float:
        """Return the component's output power in watts."""
        pass

    @property
    @abstractmethod
    def component_name(self) -> str:
        """Return a descriptive component name."""
        pass
```

### An abstract class cannot be instantiated
The following attempt raises `TypeError` because the abstract members have not
been implemented. The handler keeps this intentional demonstration from stopping
the notebook. A child that leaves either member abstract also cannot be instantiated.

```python
try:
    abstract_component = EnergyComponent()
except TypeError as error:
    print("Cannot instantiate EnergyComponent:")
    print(error)
```

### Implement a concrete solar panel
The final `SolarPanel` definition inherits from `EnergyComponent` and implements
both required members. This is a separate example from the inheritance section.

- `_model` stores the name used by the property. A leading underscore indicates
  non-public implementation detail by convention; it does not enforce privacy.
- `component_name` is accessed without parentheses. No setter is defined, so this
  property cannot be assigned through `panel.component_name`.
- `calculate_output()` is called with parentheses and returns a number.

The calculation uses the simple DC relationship **power (W) = voltage (V) × current (A)**.
It assumes valid numeric inputs and does not model losses, changing conditions,
or input validation.

```python
class SolarPanel(EnergyComponent):
    """Calculate illustrative DC output power for a solar panel."""

    def __init__(self, model: str, voltage_v: float, current_a: float) -> None:
        self._model = model
        self.voltage_v = voltage_v
        self.current_a = current_a

    @property
    def component_name(self) -> str:
        """Return the panel model as its component name."""
        return self._model

    def calculate_output(self) -> float:
        """Return DC power in watts from the stored voltage and current."""
        return self.voltage_v * self.current_a
```

```python
energy_panel = SolarPanel(
    model="Example Panel C",
    voltage_v=48.5,
    current_a=8.2,
)

# Access the property without parentheses.
print("Component:", energy_panel.component_name)

# Call the method and store its returned result.
output_power_w = energy_panel.calculate_output()
print(f"Output power: {output_power_w:.2f} W")
print("Is an EnergyComponent:", isinstance(energy_panel, EnergyComponent))
```

### Common pitfalls
- Omitting `self` from an instance method definition.
- Writing `model = model` instead of `self.model = model` when storing an attribute.
- Forgetting parentheses when calling a method, or adding them when reading a property.
- Forgetting parent initialization when the child defines its own `__init__`.
- Trying to instantiate a class that still has abstract members.
- Assuming type hints validate data or a leading underscore enforces privacy.
- Rerunning an early example after a later cell has redefined `SolarPanel`.
