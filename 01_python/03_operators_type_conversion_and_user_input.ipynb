## 1. Arithmetic operators
Arithmetic operators perform calculations on numeric values.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `+` | Addition | `7 + 3` | `10` |
| `-` | Subtraction | `7 - 3` | `4` |
| `*` | Multiplication | `7 * 3` | `21` |
| `/` | True division | `7 / 2` | `3.5` |
| `//` | Floor division | `7 // 2` | `3` |
| `%` | Modulo (remainder) | `7 % 3` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |

For built-in integers, `/` returns a float. Floor division rounds the quotient down,
not toward zero: `-7 // 2` is `-4`. Division or modulo by zero raises `ZeroDivisionError`.
Use parentheses to make the intended calculation clear.

```python
print("Addition:", 7 + 3)
print("Subtraction:", 7 - 3)
print("Multiplication:", 7 * 3)
print("True division:", 7 / 2)
print("Floor division:", 7 // 2)
print("Modulo:", 7 % 3)
print("Exponentiation:", 2 ** 3)
print("Negative floor division:", -7 // 2)
```

### Calculating power, voltage difference, and resistive loss
The following example uses voltage in volts (V), current in amperes (A), and
resistance in ohms. Power and resistive loss are expressed in watts (W).

- DC power: voltage × current.
- Voltage difference: measured voltage − reference voltage.
- Average voltage: sum of two readings divided by two.
- Resistive loss: current² × resistance.

```python
voltage_v = 48.5
second_voltage_v = 47.9
reference_voltage_v = 48.0
current_a = 8.2
resistance_ohm = 0.15

power_w = voltage_v * current_a
voltage_difference_v = voltage_v - reference_voltage_v
average_voltage_v = (voltage_v + second_voltage_v) / 2
resistive_loss_w = current_a ** 2 * resistance_ohm

print(f"DC power: {power_w:.2f} W")
print(f"Voltage difference: {voltage_difference_v:.2f} V")
print(f"Average voltage: {average_voltage_v:.2f} V")
print(f"Resistive loss: {resistive_loss_w:.2f} W")
```

In an f-string, `:.2f` displays a number with two digits after the decimal point.
It changes the displayed text, not the stored numeric value.

## 2. Assignment and augmented assignment
`=` binds a name to a value. Augmented assignment combines an operation with assignment.
For the numeric examples here, `x += 3` has the same effect as `x = x + 3`.

Each row below starts independently with `x = 10`.

| Statement | Numeric equivalent | New value of `x` |
| --- | --- | --- |
| `x += 3` | `x = x + 3` | `13` |
| `x -= 3` | `x = x - 3` | `7` |
| `x *= 3` | `x = x * 3` | `30` |
| `x /= 4` | `x = x / 4` | `2.5` |
| `x //= 3` | `x = x // 3` | `3` |
| `x %= 3` | `x = x % 3` | `1` |
| `x **= 2` | `x = x ** 2` | `100` |

### Updating accumulated energy
Add the next interval's energy, subtract a loss, and apply a correction factor.
The final variable uses a new name to make the conversion from Wh to kWh explicit.

```python
energy_wh = 1250.0
print(f"Initial energy: {energy_wh:.2f} Wh")

energy_wh += 350.0
print(f"After new interval: {energy_wh:.2f} Wh")

energy_wh -= 40.0
print(f"After estimated loss: {energy_wh:.2f} Wh")

energy_wh *= 0.98
print(f"After correction: {energy_wh:.2f} Wh")

energy_kwh = energy_wh
energy_kwh /= 1000
print(f"Converted energy: {energy_kwh:.4f} kWh")
```

```python
value = 10
value //= 3
print("Floor-divide and assign:", value)

value = 10
value %= 3
print("Modulo and assign:", value)

value = 10
value **= 2
print("Exponentiate and assign:", value)
```

## 3. Comparison operators
Comparisons of the numeric values below return `True` or `False`.
Do not confuse assignment (`=`) with equality comparison (`==`).

| Operator | Meaning |
| --- | --- |
| `==` | Equal to |
| `!=` | Not equal to |
| `<` | Less than |
| `>` | Greater than |
| `<=` | Less than or equal to |
| `>=` | Greater than or equal to |

### Comparing measurements with reference values
These checks demonstrate the six operators, not a real monitoring policy.

```python
measured_voltage_v = 48.5
reference_voltage_v = 48.0
measured_current_a = 8.2
current_limit_a = 10.0
measured_power_w = measured_voltage_v * measured_current_a
power_target_w = 380.0
temperature_c = 42.0
temperature_limit_c = 60.0
irradiance_w_m2 = 800.0
irradiance_threshold_w_m2 = 600.0

print("Voltage equals reference:", measured_voltage_v == reference_voltage_v)
print("Voltage differs from reference:", measured_voltage_v != reference_voltage_v)
print("Current below limit:", measured_current_a < current_limit_a)
print("Power exceeds target:", measured_power_w > power_target_w)
print("Temperature at or below limit:", temperature_c <= temperature_limit_c)
print("Irradiance meets threshold:", irradiance_w_m2 >= irradiance_threshold_w_m2)
```

**Floating-point note:** decimal fractions are not always represented exactly.
For example, `0.1 + 0.2 == 0.3` is `False`. Exact equality is not generally a suitable
way to compare calculated measurements; use an appropriate tolerance when needed.

## 4. Logical operators
Logical operators combine or reverse conditions.

| Operator | Meaning for Boolean operands |
| --- | --- |
| `and` | `True` when both conditions are true |
| `or` | `True` when at least one condition is true |
| `not` | Reverses the Boolean value |

`and` and `or` short-circuit: the second operand is evaluated only when needed.
They can return non-Boolean operands in general; here, all operands are Boolean conditions.

### Combining monitoring conditions
Use the measurements from the preceding section to check sunlight and temperature,
detect warnings, and account for a maintenance flag.

```python
sunlight_ok = irradiance_w_m2 >= irradiance_threshold_w_m2
temperature_ok = temperature_c <= temperature_limit_c
conditions_ok = sunlight_ok and temperature_ok

high_current = measured_current_a > current_limit_a
high_temperature = temperature_c > temperature_limit_c
has_warning = high_current or high_temperature

maintenance_required = False
maintenance_clear = not maintenance_required
ready_to_operate = conditions_ok and maintenance_clear and (not has_warning)

print("Sunlight and temperature acceptable:", conditions_ok)
print("At least one warning:", has_warning)
print("Maintenance clear:", maintenance_clear)
print("Ready under the example rules:", ready_to_operate)
```

## 5. Membership operators
`in` checks whether an item is present; `not in` checks whether it is absent.
For strings, membership checks for a substring and is case-sensitive.
For dictionaries, membership checks keys by default.

### Checking panel labels and a status message
Lowercasing the message makes the lowercase keyword match regardless of the
capitalization used in this example.

```python
panel_labels = ["PV_01", "PV_02", "PV_03"]
status_message = "WARNING: High panel temperature"

print("PV_02 is present:", "PV_02" in panel_labels)
print("PV_04 is absent:", "PV_04" not in panel_labels)
print("Lowercase keyword in original message:", "warning" in status_message)
print("Lowercase keyword in normalized message:", "warning" in status_message.lower())
```

## 6. Identity and equality
- `==` compares values.
- `is` checks whether two names refer to the same object.
- `is not` checks whether they refer to different objects.

Equal lists can be separate objects. Assigning an existing list to another name
creates a shared reference, not a copy. Use `==`, not `is`, to compare numeric or text values.

```python
panels_a = ["PV_01", "PV_02"]
panels_b = ["PV_01", "PV_02"]
panels_shared = panels_a

print("Equal values:", panels_a == panels_b)
print("Same object:", panels_a is panels_b)
print("Different objects:", panels_a is not panels_b)
print("Shared reference:", panels_a is panels_shared)
```

### Updating a shared list
Appending through `panels_shared` also changes what `panels_a` refers to.
`panels_b` remains unchanged because it is a separate list.
Re-run the preceding initialization cell before repeating this update.

```python
panels_shared.append("PV_03")

print("panels_a:", panels_a)
print("panels_b:", panels_b)
print("panels_shared:", panels_shared)
print("Still the same object:", panels_a is panels_shared)
print("Still equal to panels_b:", panels_a == panels_b)
```

### Checking for a missing value
Use `is None` or `is not None` to check for `None`.

```python
missing_voltage = None

print("Voltage is missing:", missing_voltage is None)
print("Voltage is available:", missing_voltage is not None)
```

## 7. Type conversion
**Implicit conversion** happens automatically in supported operations.
For example, adding an `int` and a `float` produces a `float`.

**Explicit conversion** uses a function such as `int()`, `float()`, `str()`, or `bool()`.
A conversion returns a value; it does not change the original variable unless you assign the result.

```python
whole_value = 3
decimal_value = 2.5
combined_value = whole_value + decimal_value

print("Implicit conversion result:", combined_value, type(combined_value))
print("Original integer:", whole_value, type(whole_value))
```

### Converting measurement data
| Function | Purpose | Example |
| --- | --- | --- |
| `int()` | Convert to an integer | `int("12")` → `12` |
| `float()` | Convert to a floating-point number | `float("48.5")` → `48.5` |
| `str()` | Convert to text | `str(12)` → `"12"` |
| `bool()` | Evaluate truthiness | `bool(0)` → `False` |

`int()` truncates a finite float toward zero; it does not round to the nearest integer.
Numeric text must have a valid format: `int("hello")` and `int("3.5")` raise `ValueError`.
Use `float("3.5")` when the text represents a decimal value.

```python
panel_count_text = "12"
voltage_text = "48.5"
measured_temperature_c = 42.8

panel_count = int(panel_count_text)
converted_voltage_v = float(voltage_text)
integer_temperature_c = int(measured_temperature_c)
panel_count_as_text = str(panel_count)
reading_is_nonzero = bool(converted_voltage_v)

print("Panel count:", panel_count, type(panel_count))
print("Voltage:", converted_voltage_v, type(converted_voltage_v))
print("Integer temperature:", integer_temperature_c, type(integer_temperature_c))
print("Panel count as text:", panel_count_as_text, type(panel_count_as_text))
print("Nonzero reading:", reading_is_nonzero, type(reading_is_nonzero))
print("Negative truncation:", int(-3.9))
```

### Truthiness is not text interpretation
Zero, empty strings, empty collections, and `None` are falsy.
Non-empty strings are truthy, including `"False"` and `"0"`.
Thus, `bool()` does not parse a textual Boolean answer.

```python
nonempty_text = bool("Ready")
empty_text = bool("")

print("Non-empty string:", nonempty_text, type(nonempty_text))
print("Empty string:", empty_text, type(empty_text))
print('Text "False":', bool("False"))
print('Text "0":', bool("0"))
print("Numeric zero:", bool(0))
print("Empty list:", bool([]))
print("None:", bool(None))
```

## 9. Number bases and bits
A **bit** is `0` or `1`; one **byte** contains eight bits.
Different number bases can represent the same integer.

| Base | Name | Digits | Python integer literal |
| --- | --- | --- | --- |
| 10 | Decimal | `0`–`9` | `10` |
| 2 | Binary | `0`, `1` | `0b1010` |
| 16 | Hexadecimal | `0`–`9`, `A`–`F` | `0xA` |

Binary position weights are powers of two:
`0b1010 = 1 × 2³ + 0 × 2² + 1 × 2¹ + 0 × 2⁰ = 10`.

`bin()` and `hex()` return **strings** for display. `int(text, base)` parses a string
using the specified base. Python accepts lowercase hexadecimal digits as well.

```python
decimal_value = 10
binary_value = 0b1010
hexadecimal_value = 0xA

print("Same integer:", decimal_value == binary_value == hexadecimal_value)
print("Binary literal value and type:", binary_value, type(binary_value))
print("Binary display:", bin(decimal_value), type(bin(decimal_value)))
print("Hexadecimal display:", hex(decimal_value), type(hex(decimal_value)))
print("Parsed binary:", int("1010", 2))
print("Parsed hexadecimal:", int("A", 16))
```

## 10. Bitwise operators
Bitwise operators work on the binary representation of integers.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `&` | AND: bit is 1 when both bits are 1 | `6 & 3` | `2` |
| `\|` | OR: bit is 1 when either bit is 1 | `6 \| 3` | `7` |
| `^` | XOR: bit is 1 when the bits differ | `6 ^ 3` | `5` |
| `~` | Bitwise inversion | `~6` | `-7` |
| `<<` | Left shift | `6 << 1` | `12` |
| `>>` | Right shift | `6 >> 1` | `3` |

For Python integers, `~x == -x - 1`. Integers are not limited to an eight-bit width,
so inversion should not be read as simply flipping a displayed byte.
For a non-negative integer shift count `n`, `x << n` equals `x * 2 ** n`, and
`x >> n` equals `x // 2 ** n`.

**Do not substitute bitwise operators for `and`, `or`, and `not` in Boolean conditions.**

### Combining example status codes
Use `6` (`0b110`) and `3` (`0b011`) as simplified status codes.
The example demonstrates bit patterns only; it does not assign operational meanings to the bits.

```python
status_a = 6
status_b = 3

print("First status, decimal:", status_a)
print("First status, binary:", bin(status_a))
print("First status, hexadecimal:", hex(status_a))
print("Second status, binary:", bin(status_b))

and_result = status_a & status_b
or_result = status_a | status_b
xor_result = status_a ^ status_b

print("AND:", and_result, bin(and_result))
print("OR:", or_result, bin(or_result))
print("XOR:", xor_result, bin(xor_result))
```

### Inverting and shifting bits
`bin()` displays a negative value with a minus sign; it does not display a fixed-width
two's-complement bit pattern.

```python
inverted_status = ~status_a
left_shifted_status = status_a << 1
right_shifted_status = status_a >> 1

print("Inversion:", inverted_status, bin(inverted_status))
print("Left shift:", left_shifted_status, bin(left_shifted_status))
print("Right shift:", right_shifted_status, bin(right_shifted_status))
```
