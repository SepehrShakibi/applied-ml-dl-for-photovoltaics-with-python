## 1. Function structure
A function is a reusable block of code designed to perform a specific task.
Defining a function makes it available; its body runs only when the function is called.

| Component | Purpose |
| --- | --- |
| `def` | Start a function definition |
| Function name | Identify the operation; use descriptive `snake_case` names |
| Parameters | Name the inputs in the function definition |
| Arguments | Supply actual values when calling the function |
| `:` and indentation | Introduce and delimit the function body |
| `return` | End the function call and send a value back to the caller |

A **docstring** is a string at the beginning of the body that documents the function.
Comments explain implementation details when needed. Use four spaces for indentation.

## 2. Returning a result and performing an action
`return` and `print()` have different purposes:

- `return` sends a value to the caller for further use.
- `print()` displays text; it does not substitute for returning a calculated value.

A function that reaches the end without an explicit `return` returns `None`.
A bare `return` also returns `None`. Thus, an action-only function still has a
return value, but it provides no meaningful calculated result.

## 4. Type hints
Type hints are optional annotations describing the expected types of inputs and results.
They improve documentation and support editor suggestions and static analysis.

- `parameter: float` annotates a parameter.
- `-> float` annotates a numeric return value.
- `-> None` describes a function with no meaningful return result.
- `-> tuple[float, str]` describes a tuple containing a float followed by a string.

**Type hints do not automatically convert or validate values at runtime.**
A string received from `input()` still needs explicit conversion with `float()`.

## 5. Errors and exceptions
**Error** is a general term for a problem in a program. An **exception** interrupts
the normal flow of execution and can often be handled by the program.

A syntax error prevents the affected script or notebook cell from being parsed
successfully. A runtime exception occurs while validly parsed code executes.
A logical error may produce an incorrect result without raising any exception.

| Example | Exception |
| --- | --- |
| `float("invalid")` | `ValueError`: invalid text for numeric conversion |
| `10.0 / 0.0` | `ZeroDivisionError`: division by zero |
| Opening a nonexistent file for reading | `FileNotFoundError` |
| `"48.5" + 1.0` | `TypeError`: unsupported operand types |

## 6. Handling specific exceptions
Put the operation that may fail inside `try`, then handle an expected exception
with `except ExceptionType:`.

- If no exception occurs, Python skips the handler.
- If a matching exception occurs, Python skips the remaining statements in the
  `try` block and runs the handler.
- Execution continues after the statement when the exception is handled.
- An exception with no matching handler propagates to the caller; if unhandled,
  it stops the current cell and displays a traceback.

Place the specific `ValueError` handler **before** the general `Exception` handler.
Python uses the first matching handler. Because `ValueError` is a subclass of
`Exception`, placing the general handler first would hide the specific handling.

`except Exception as error` binds the caught exception to `error`, allowing its
type and message to be displayed. It catches most ordinary runtime exceptions,
but not `KeyboardInterrupt` or `SystemExit`.

## Summary
| Concept | Key point |
| --- | --- |
| Function definition | Use `def`, a descriptive name, parameters, and an indented body |
| Function call | Supply arguments to execute the function body |
| `return` | End a call and send a result to the caller |
| Action-only function | Performs an action and implicitly returns `None` |
| Multiple results | Return a tuple and unpack its items in order |
| Type hints | Document expected types; do not enforce runtime validation |
| `try` / `except` | Handle matching exceptions raised by operations in `try` |
| Specific handlers | Provide targeted responses to expected failures |
| General fallback | Place `Exception` after specific handlers and report the problem |

### Common pitfalls
- Defining a function but forgetting to call it.
- Confusing printed output with a returned value.
- Using a function's local variable outside that function instead of using its return value.
- Unpacking a tuple into the wrong number of variables.
- Assuming type hints automatically convert user input.
- Putting numeric conversion outside the `try` block intended to handle it.
- Placing `except Exception` before `except ValueError`.
