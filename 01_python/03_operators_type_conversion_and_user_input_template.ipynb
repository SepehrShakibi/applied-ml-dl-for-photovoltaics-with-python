## 1. Arithmetic operators
Arithmetic operators perform calculations on numeric values.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `+` | Addition | `7 + 3` | `10` |
| `-` | Subtraction | `7 - 3` | `4` |
| `*` | Multiplication | `7 * 3` | `21` |
| `/` | True division | `7 / 2` | `3.5` |
| `//` | Floor division | `7 // 2` | `3` |
| `%` | Modulo (remainder) | `7 % 3` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |

For built-in integers, `/` returns a float. Floor division rounds the quotient down,
not toward zero: `-7 // 2` is `-4`. Division or modulo by zero raises `ZeroDivisionError`.
Use parentheses to make the intended calculation clear.

### Calculating power, voltage difference, and resistive loss

- DC power: voltage × current.

| `x *= 3` | `x = x * 3` | `30` |

| `x %= 3` | `x = x % 3` | `1` |
| `x **= 2` | `x = x ** 2` | `100` |

## 6. Identity and equality
- `==` compares values.
- `is` checks whether two names refer to the same object.
- `is not` checks whether they refer to different objects.

Equal lists can be separate objects. Assigning an existing list to another name
creates a shared reference, not a copy. Use `==`, not `is`, to compare numeric or text values.

```python
panels_a = ["PV_01", "PV_02"]
panels_b = ["PV_01", "PV_02"]
panels_shared = panels_a

print("Equal values:", panels_a == panels_b)
print("Same object:", panels_a is panels_b)
print("Different objects:", panels_a is not panels_b)
print("Shared reference:", panels_a is panels_shared)
```

### Updating a shared list

## 7. Type conversion
**Implicit conversion** happens automatically in supported operations.
For example, adding an `int` and a `float` produces a `float`.

**Explicit conversion** uses a function such as `int()`, `float()`, `str()`, or `bool()`.
A conversion returns a value; it does not change the original variable unless you assign the result.

### Converting measurement data
| Function | Purpose | Example |
| --- | --- | --- |
| `int()` | Convert to an integer | `int("12")` → `12` |
| `float()` | Convert to a floating-point number | `float("48.5")` → `48.5` |
| `str()` | Convert to text | `str(12)` → `"12"` |
| `bool()` | Evaluate truthiness | `bool(0)` → `False` |

`int()` truncates a finite float toward zero; it does not round to the nearest integer.
Numeric text must have a valid format: `int("hello")` and `int("3.5")` raise `ValueError`.
Use `float("3.5")` when the text represents a decimal value.

### Truthiness is not text interpretation
Zero, empty strings, empty collections, and `None` are falsy.
Non-empty strings are truthy, including `"False"` and `"0"`.
Thus, `bool()` does not parse a textual Boolean answer.

## 9. Number bases and bits
A **bit** is `0` or `1`; one **byte** contains eight bits.
Different number bases can represent the same integer.

| Base | Name | Digits | Python integer literal |
| --- | --- | --- | --- |
| 10 | Decimal | `0`–`9` | `10` |
| 2 | Binary | `0`, `1` | `0b1010` |
| 16 | Hexadecimal | `0`–`9`, `A`–`F` | `0xA` |

Binary position weights are powers of two:
`0b1010 = 1 × 2³ + 0 × 2² + 1 × 2¹ + 0 × 2⁰ = 10`.

`bin()` and `hex()` return **strings** for display. `int(text, base)` parses a string
using the specified base. Python accepts lowercase hexadecimal digits as well.

## 10. Bitwise operators
Bitwise operators work on the binary representation of integers.

| Operator | Meaning | Example | Result |
| --- | --- | --- | --- |
| `&` | AND: bit is 1 when both bits are 1 | `6 & 3` | `2` |
| `\|` | OR: bit is 1 when either bit is 1 | `6 \| 3` | `7` |
| `^` | XOR: bit is 1 when the bits differ | `6 ^ 3` | `5` |
| `~` | Bitwise inversion | `~6` | `-7` |
| `<<` | Left shift | `6 << 1` | `12` |
| `>>` | Right shift | `6 >> 1` | `3` |

For Python integers, `~x == -x - 1`. Integers are not limited to an eight-bit width,
so inversion should not be read as simply flipping a displayed byte.
For a non-negative integer shift count `n`, `x << n` equals `x * 2 ** n`, and
`x >> n` equals `x // 2 ** n`.

**Do not substitute bitwise operators for `and`, `or`, and `not` in Boolean conditions.**
