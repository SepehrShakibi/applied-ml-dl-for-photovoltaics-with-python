## 4. Working with strings
Strings are ordered sequences of characters.

- Indexing starts at `0`; `-1` selects the last character.
- Slicing uses `[start:stop]`; the stop index is excluded.
- `len()` returns the number of characters.
- Strings are immutable: methods return new strings rather than changing the original.

## 5. None
`None` represents the absence of a value. It is not the same as `0` or `False`.
Use `is None` to check whether a value is `None`.

## 6. One-dimensional lists
Lists are ordered and mutable. They allow duplicates and can contain different types,
including other collections. Indexing and slicing follow the same rules as strings.

### Creating and editing a list
The following example accesses and slices a list of voltage readings, then adds,
updates, and removes items before displaying the final list.

### Other list operations
- `insert(index, value)` inserts an item.
- `remove(value)` removes the first matching value.
- `pop(index)` removes and returns an item; the default is the last item.
- `sort()` sorts the list in place.
- `reverse()` reverses the list in place.

`remove()` requires an existing value, and indexed operations require a valid index.

## 7. Multidimensional lists
A two-dimensional list is a list of rows: `table[row][column]`.
A three-dimensional list contains layers of rows: `cube[layer][row][column]`.
All indices start at `0`.

## 8. Dictionaries
Dictionaries store key-value pairs. Keys are unique and must be hashable;
strings are common keys. Values can have different types.
Dictionaries are mutable and accessed by key rather than position.

### Other dictionary operations
- `get(key, default)` reads a value without raising `KeyError` for a missing key.
- `update()` adds or updates multiple pairs.
- `pop(key)` removes a key and returns its value.

Direct access with `system[key]` raises `KeyError` if the key does not exist.

## 9. Tuples and sets
A **tuple** is ordered and immutable: its items cannot be reassigned.
It supports indexing and allows duplicates.

A **set** stores unique, hashable elements. It is unordered and does not support indexing.
Sets are useful for membership testing and removing duplicates.

## Summary
| Collection | Access | Mutable? | Duplicates? |
| --- | --- | --- | --- |
| List | Index | Yes | Yes |
| Tuple | Index | No | Yes |
| Dictionary | Key | Yes | Keys: no; values: yes |
| Set | No indexing | Yes | No |

Set display order may vary. Use `set()` for an empty set; `{}` creates an empty dictionary.
