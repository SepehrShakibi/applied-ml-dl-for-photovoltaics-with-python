## 1. Variables and data types
A variable is a name that refers to an object. Use descriptive `snake_case` names.
Names are case-sensitive, cannot start with a digit, and cannot be Python keywords.

| Type | Meaning | Example |
| --- | --- | --- |
| `int` | Integer | `12` |
| `float` | Floating-point number | `48.5` |
| `str` | Text | `"Solar Lab"` |
| `bool` | Boolean value | `True` |

### Creating variables
The following example defines a system identifier, module count, voltage, current, operating status, and note.
`type()` shows the type of each value.

```python
system_id = "PV_01"
module_count = 12
voltage = 48.5
current = 8.2
is_operational = True
note = "Example system for Python practice."

print(system_id, type(system_id))
print(module_count, type(module_count))
print(voltage, type(voltage))
print(current, type(current))
print(is_operational, type(is_operational))
print(note, type(note))
```

## 4. Working with strings
Strings are ordered sequences of characters.

- Indexing starts at `0`; `-1` selects the last character.
- Slicing uses `[start:stop]`; the stop index is excluded.
- `len()` returns the number of characters.
- Strings are immutable: methods return new strings rather than changing the original.

```python
label = "Python_2026!"

print(len(label))
print(label[0])
print(label[-1])
print(label[0:6])
print(label.upper())
print(label.lower())
print(label.replace("2024", "2025"))
print(label.split("_"))
print("  Python  ".strip())
print("Original:", label)
```

### Cleaning text data
This workflow removes extra spaces, normalizes case, replaces an unwanted character,
splits structured text, and extracts meaningful segments before printing the results.

```python
raw_site = "  Solar-Lab  "
raw_sensor = "  INV01_TEMP  "
raw_file = "  PV_01_2025.csv  "

site_name = raw_site.strip().lower().replace("-", "_")
sensor_id = raw_sensor.strip().upper()
sensor_parts = sensor_id.split("_")
measurement_label = sensor_parts[1]
file_name = raw_file.strip().lower()
file_stem = file_name[:-4]  # Remove the known .csv suffix.

print(f"Site: {site_name}")
print(f"Sensor: {sensor_id}")
print(f"Sensor parts: {sensor_parts}")
print(f"Measurement: {measurement_label}")
print(f"File: {file_name}")
print(f"File stem: {file_stem}")
```

## 5. None
`None` represents the absence of a value. It is not the same as `0` or `False`.
Use `is None` to check whether a value is `None`.

```python
missing_voltage = None
zero_voltage = 0.0
inactive_status = False

print("Missing:", missing_voltage, type(missing_voltage))
print("Zero:", zero_voltage, type(zero_voltage))
print("Inactive:", inactive_status, type(inactive_status))
print(missing_voltage is None)
print(zero_voltage is None)
print(inactive_status is None)
```

## 6. One-dimensional lists
Lists are ordered and mutable. They allow duplicates and can contain different types,
including other collections. Indexing and slicing follow the same rules as strings.

### Creating and editing a list
The following example accesses and slices a list of voltage readings, then adds,
updates, and removes items before displaying the final list.

```python
voltages = [48.1, 48.5, 48.3, 48.5]

print("Initial:", voltages)
print("First:", voltages[0])
print("Last:", voltages[-1])
print("First two:", voltages[:2])

voltages.append(48.7)
voltages[1] = 48.6
voltages.remove(48.3)

print("Final:", voltages)
print("Count:", len(voltages))
```

### Other list operations
- `insert(index, value)` inserts an item.
- `remove(value)` removes the first matching value.
- `pop(index)` removes and returns an item; the default is the last item.
- `sort()` sorts the list in place.
- `reverse()` reverses the list in place.

`remove()` requires an existing value, and indexed operations require a valid index.

```python
readings = [48.5, 48.1, 48.3]
readings.insert(1, 48.2)
print("After insert:", readings)

removed_reading = readings.pop(0)
print("Removed:", removed_reading)

readings.sort()
print("Sorted:", readings)

readings.reverse()
print("Reversed:", readings)
```

## 7. Multidimensional lists
A two-dimensional list is a list of rows: `table[row][column]`.
A three-dimensional list contains layers of rows: `cube[layer][row][column]`.
All indices start at `0`.

### Working with rows, columns, and layers
The 2D example demonstrates accessing a row and an item, updating an item, and adding and removing a row.
The 3D example demonstrates accessing data by layer, row, and column.

In these examples, rows represent sensors, columns represent time steps,
and layers represent sites. Values are example voltages in volts.

```python
voltage_table = [
    [48.1, 48.2, 48.3],
    [47.9, 48.0, 48.1],
]

print("First row:", voltage_table[0])
print("Second row, third column:", voltage_table[1][2])

voltage_table[1][2] = 48.4
voltage_table.append([48.5, 48.6, 48.7])
print("After update and append:", voltage_table)

removed_row = voltage_table.pop()
print("Removed row:", removed_row)
print("Final table:", voltage_table)
```

```python
voltage_cube = [
    [
        [48.1, 48.2, 48.3],
        [47.9, 48.0, 48.1],
    ],
    [
        [49.1, 49.2, 49.3],
        [48.9, 49.0, 49.1],
    ],
]

print("Second layer:", voltage_cube[1])
print("Second layer, first row:", voltage_cube[1][0])
print("Second layer, first row, third column:", voltage_cube[1][0][2])
```

### Reading the structure
For `voltage_cube`:

| Dimension | Meaning | Size |
| --- | --- | --- |
| Layer | Site | 2 |
| Row | Sensor | 2 per site |
| Column | Time step | 3 per sensor |

`voltage_cube[1][0][2]` selects `49.3`.
Nested lists do not require equal row lengths; these examples use a regular layout.

## 8. Dictionaries
Dictionaries store key-value pairs. Keys are unique and must be hashable;
strings are common keys. Values can have different types.
Dictionaries are mutable and accessed by key rather than position.

### Building a system record
This record contains site, equipment, measurement, and operating-status information.
The examples demonstrate adding a key, updating a value, inspecting keys and values,
and iterating through key-value pairs.

```python
system = {
    "site_name": "Solar Lab",
    "system_id": "PV_01",
    "module_count": 12,
    "voltage": 48.5,
    "current": 8.2,
    "is_operational": True,
}

print("Site:", system["site_name"])

system["note"] = "Example record."
system["voltage"] = 48.7

print("Keys:", system.keys())
print("Values:", system.values())
print("Pairs:", system.items())
```

### Iterate through key-value pairs
This small `for` loop prints one pair at a time.
`items()` provides each key and its value. The indented line is repeated for every pair.
Loops will be covered in more detail later.

```python
for key, value in system.items():
    print(f"{key}: {value}")
```

### Other dictionary operations
- `get(key, default)` reads a value without raising `KeyError` for a missing key.
- `update()` adds or updates multiple pairs.
- `pop(key)` removes a key and returns its value.

Direct access with `system[key]` raises `KeyError` if the key does not exist.

```python
print("Missing temperature:", system.get("temperature"))
print("Missing operator:", system.get("operator", "Not assigned"))

system.update({"current": 8.4, "status": "Running"})
removed_note = system.pop("note")

print("Removed note:", removed_note)
print("Updated record:", system)
```

## 9. Tuples and sets
A **tuple** is ordered and immutable: its items cannot be reassigned.
It supports indexing and allows duplicates.

A **set** stores unique, hashable elements. It is unordered and does not support indexing.
Sets are useful for membership testing and removing duplicates.

### Creating a tuple and a set
The tuple contains three panel labels. The set is created with a repeated label,
which is stored only once. The output shows indexing, collection lengths, and membership testing.

```python
panel_tuple = ("PV_01", "PV_02", "PV_03")
panel_set = {"PV_01", "PV_02", "PV_01"}

print("Tuple:", panel_tuple)
print("First tuple item:", panel_tuple[0])
print("Tuple length:", len(panel_tuple))

print("Set:", panel_set)
print("Set length:", len(panel_set))
print("PV_02 is in the set:", "PV_02" in panel_set)
```

## Summary
| Collection | Access | Mutable? | Duplicates? |
| --- | --- | --- | --- |
| List | Index | Yes | Yes |
| Tuple | Index | No | Yes |
| Dictionary | Key | Yes | Keys: no; values: yes |
| Set | No indexing | Yes | No |

Set display order may vary. Use `set()` for an empty set; `{}` creates an empty dictionary.
